### 2. Regression for Predicting Goals

Regression is typically used for continuous outcomes, while predicting whether a shot results in a goal is a binary classification problem. Classification models like logistic regression, decision trees, or ensemble methods are more suitable as they are designed to handle categorical outcomes and provide probabilities for class memberships.

### 4. K-Fold Cross-Validation

#### Advantages
- Bias-Variance Trade-off: Balances bias and variance by training and validating on different subsets.
- Efficient Use of Data: Utilizes all data for training and validation.

#### Disadvantages
- Computationally Expensive: Requires multiple model trainings.
- Not Suitable for Time Series: Doesn't account for temporal dependencies.

#### Alternative Methods
- Stratified K-Fold: Ensures each fold has a proportional representation of each class.
- Leave-One-Out Cross-Validation (LOOCV): Uses one data point as the validation set and the rest as the training set.
- Time Series Split: Ensures the training set always precedes the validation set, maintaining temporal order.
- Repeated K-Fold: Repeats K-Fold cross-validation multiple times to reduce variability.

### 5. Metrics for Evaluating Manual Implementations

#### Metrics Used
- Mean Squared Error (MSE): Measures the average squared difference between predicted and actual values.
- R-Squared (R²): Indicates the proportion of variance explained by the model.

#### Rationale
- MSE: Provides a clear measure of error magnitude and is sensitive to large errors.
- R²: Offers insight into the model's explanatory power and effectiveness.
